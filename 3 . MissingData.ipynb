```python
import numpy as np
import pandas as pd
```

### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df.isna().any()       ## to check is there any coulmn with no null values
```

### Removing Missing Values

```python
df
```

```python
df.dropna()       ## null value rows removed
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df1 = pd.DataFrame(data)
df1
```

```python
df1.dropna()
```

```python
df1.dropna(thresh = 3)       ## atleast 3 no null values in row 
```
