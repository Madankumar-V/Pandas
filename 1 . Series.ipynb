### Pandas Series

A **Series** is a one-demnsional labeled array capable of holding any data type . The axis labels are collectively called the index.

### Creating Series

There are multiple ways to create a pandas series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(my_list)       ## first column -- index (labels)
                         ## second column  -- values(data)
```

```python
pd.Series(my_list , index = labels)  ## custom labels or index     
```

```python
pd.Series(arr)         ## arr should be 1d , bcz series itself is 1d
```

```python
pd.Series(d)    ## no custom indexes , auto created with dict
```
